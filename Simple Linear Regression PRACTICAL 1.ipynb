```python
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
sns.scatterplot(x=npd['daily'],y=npd['sunday'])
plt.xlabel("daily")
plt.ylabel("sunday")
plt.show()
```

```python
x=npd[['daily']]
x
```

```python
y=npd['sunday']
y
```

```python
from sklearn.linear_model import LinearRegression
```

```python
slr = LinearRegression()#model building
slr.fit(x,y)# model training
y_pred = slr.predict(x).round(2) # model testing
```

```python
y_pred.round(2)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error,mean_absolute_percentage_error,root_mean_squared_error
```

```python
slr.coef_
```

```python
slr.intercept_
```

```python
r2_score(y,y_pred)
```

```python
mean_squared_error(y,y_pred)
```

```python
mean_absolute_error(y,y_pred)
```

```python
mean_absolute_percentage_error(y,y_pred)
```

```python
root_mean_squared_error(y,y_pred)
```
